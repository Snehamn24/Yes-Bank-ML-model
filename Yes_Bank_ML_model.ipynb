# **Project Summary -**

The objective of this project is to predict the monthly closing stock price of Yes Bank using machine learning regression techniques. Yes Bank has been one of the most discussed banks in the Indian financial sector, especially after the fraud case involving its founder Rana Kapoor in 2018. Due to this major financial controversy, the bank experienced significant fluctuations in its stock prices, making it an interesting real-world use case for predictive analysis and time-series-related regression modeling.

The dataset used in this project contains monthly stock price information of Yes Bank from its inception. The dataset includes features such as opening price, highest price, lowest price, and closing price for each month. The target variable of this project is the “Close” column, which represents the closing stock price of the month. The main goal is to build a machine learning model capable of accurately predicting the closing price based on the other stock price features.

The project begins with understanding and exploring the dataset using various data analysis techniques. Important Python libraries such as Pandas and NumPy are used for data cleaning, manipulation, and preprocessing. The dataset is checked for duplicate values, missing values, data types, and statistical summaries to ensure data quality before model training.

Extensive exploratory data analysis (EDA) is performed using visualization libraries such as Matplotlib and Seaborn. Multiple univariate, bivariate, and multivariate charts are created to understand the relationships among variables. These visualizations help identify trends, correlations, stock price volatility, and the impact of financial events on Yes Bank’s stock performance. Correlation analysis shows that the opening, highest, and lowest prices are strongly related to the closing price, indicating that these variables can effectively contribute to prediction modeling.

After preprocessing and visualization, machine learning models are developed using Scikit-learn. Different regression algorithms such as Linear Regression, Decision Tree Regressor, Random Forest Regressor, and other suitable models are implemented and compared. The dataset is split into training and testing sets to evaluate model performance fairly. Performance metrics including R² Score, Mean Absolute Error (MAE), Mean Squared Error (MSE), and Root Mean Squared Error (RMSE) are used to measure prediction accuracy.

Further improvements are achieved through cross-validation and hyperparameter tuning techniques. These optimization methods help improve model stability and reduce overfitting. Among all models, ensemble-based methods such as Random Forest generally provide better prediction accuracy because they can capture non-linear relationships in stock price data more effectively.

The final outcome of this project demonstrates how machine learning can be used in the financial domain to predict stock prices based on historical market data. The project also highlights the importance of data analysis and visualization in understanding business trends and financial risk. This predictive system can assist investors, analysts, and financial institutions in making data-driven investment decisions and evaluating stock market behavior more efficiently.
