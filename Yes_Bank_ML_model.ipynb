### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):

There is no significant relationship between Open price and Close price.

Alternate Hypothesis (H₁):

There is a significant relationship between Open price and Close price.

#### 2. Perform an appropriate statistical test.

````python
# ==========================================
# Hypothesis Testing - 1
# Open Price vs Close Price
# ==========================================

from scipy.stats import pearsonr

# Pearson Correlation Test
correlation, p_value = pearsonr(df_copy['Open'], df_copy['Close'])

print("Correlation Coefficient :", correlation)
print("P-Value :", p_value)

# Conclusion
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test was performed.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Boxplot for detecting outliers

plt.figure(figsize=(12,6))

sns.boxplot(data=df_copy[['Open','High','Low','Close']])

plt.title('Boxplot for Outlier Detection')

plt.show()
````

````python
# ==========================================
# Outlier Treatment using IQR Method
# ==========================================

# Creating copy
df_outlier = df_copy.copy()

# Numerical columns
columns = ['Open','High','Low','Close']

# Removing Outliers
for col in columns:

    Q1 = df_outlier[col].quantile(0.25)
    Q3 = df_outlier[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_outlier = df_outlier[
        (df_outlier[col] >= lower_bound) &
        (df_outlier[col] <= upper_bound)
    ]

print("Outlier Treatment Completed")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The Interquartile Range (IQR) method was used for detecting and handling outliers.

Why IQR Method was used:
It is one of the most effective techniques for detecting outliers in numerical datasets.
Stock price data may contain extreme values due to sudden market fluctuations, fraud cases, or economic events.
Removing extreme outliers helps improve machine learning model performance and reduces prediction errors.
IQR method is simple, robust, and works well even when data is not normally distributed.
